```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
df = pd.DataFrame({
    'city': rng.choice(['SF', 'NYC', 'Austin', 'Seattle'], 500),
    'temp': rng.normal(18, 6, 500).round(1),
    'humidity': rng.uniform(30, 90, 500).round(1),
})
df.head(10)
```

```python
df.groupby('city').agg(mean_temp=('temp', 'mean'), mean_humidity=('humidity', 'mean'))
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 4))
for city, group in df.groupby('city'):
    ax.scatter(group['temp'], group['humidity'], label=city, alpha=0.6, s=18)
ax.set_xlabel('Temperature (°C)')
ax.set_ylabel('Humidity (%)')
ax.set_title('Temperature vs. humidity by city')
ax.legend()
plt.show()
```
