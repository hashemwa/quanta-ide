# A little data. A clearer picture.

Explore a synthetic dataset with Python, native tables, and inline plots.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
weekly = pd.DataFrame({
    "week": pd.date_range("2025-01-06", periods=52, freq="W-MON"),
    "orders": np.linspace(120, 420, 52).astype(int) + rng.integers(-15, 15, 52),
    "new_customers": rng.integers(40, 90, 52),
    "returns": rng.integers(0, 8, 52),
    "avg_order": rng.uniform(38, 46, 52).round(2),
})
weekly["revenue"] = (weekly["orders"] * weekly["avg_order"]).round(2)
weekly.tail(5)
```

## Follow the trend

A shared Python session keeps your data ready for the next question.

```python
ax = weekly.plot(
    x="week", y="revenue", figsize=(12.5, 4.2),
    color="#818cf8", linewidth=2.5, legend=False,
    title="A year of growth", xlabel="", ylabel="Revenue ($)",
)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.show()
```
