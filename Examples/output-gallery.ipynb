## 8 · Matplotlib figures *(needs matplotlib)*

````python
import matplotlib.pyplot as plt

t = np.linspace(0, 10, 300)
fig, ax = plt.subplots(figsize=(7, 3.2))
ax.plot(t, np.sin(t), label="sin")
ax.plot(t, np.cos(t), label="cos")
ax.set_title("A figure still open when the cell ends")
ax.legend();
````

````python
for power in (1, 2):
    plt.figure(figsize=(5, 2.4))
    plt.plot(t, t ** power)
    plt.title(f"plt.show() call {power} emits its own figure")
    plt.show()
    print(f"Text printed after figure {power}")
````

````python
fig, axes = plt.subplots(1, 3, figsize=(10, 3))
rng = np.random.default_rng(1)
axes[0].hist(rng.normal(size=1000), bins=30)
axes[1].scatter(rng.random(80), rng.random(80), c=rng.random(80))
axes[2].imshow(rng.random((20, 20)), cmap="viridis")
for ax, title in zip(axes, ["histogram", "scatter", "image"]):
    ax.set_title(title)
fig.tight_layout()
````

````python
class Badge:
    def _repr_svg_(self):
        return ('<svg xmlns="http://www.w3.org/2000/svg" width="220" height="60">'
                '<rect width="220" height="60" rx="12" fill="#4c6ef5"/>'
                '<text x="110" y="37" font-size="20" text-anchor="middle" fill="white" '
                'font-family="-apple-system">SVG output</text></svg>')

Badge()
````

````python
import struct
import zlib

class Gradient:
    def _repr_png_(self, width=240, height=80):
        rows = b"".join(b"\x00" + bytes(channel for x in range(width)
                                         for channel in (x * 255 // width, y * 255 // height, 180))
                        for y in range(height))
        def chunk(tag, data):
            return struct.pack(">I", len(data)) + tag + data + struct.pack(">I", zlib.crc32(tag + data) & 0xFFFFFFFF)
        header = struct.pack(">IIBBBBB", width, height, 8, 2, 0, 0, 0)
        return b"\x89PNG\r\n\x1a\n" + chunk(b"IHDR", header) + chunk(b"IDAT", zlib.compress(rows)) + chunk(b"IEND", b"")

Gradient()
````

## 16 · Asynchronous code

````python
import asyncio

async def fetch(delay, value):
    await asyncio.sleep(delay)
    return value

await asyncio.gather(fetch(0.2, "top-level"), fetch(0.1, "await"), fetch(0.05, "works"))
````
